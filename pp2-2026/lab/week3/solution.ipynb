# 주소록 작성

-딕셔너리를 사용해 연락처를 저장한다.
-사용자에게 메뉴를 제시하여 연락처 추가, 삭제, 검색을 지원

```python
def main():
    address_book = {}

    while True:
        user = display_menu()

        if user == 1:
            name, number = get_contact()
            address_book[name] = number
            print(f"{name}님의 연락처가 추가되었습니다.")

        elif user == 2:
            name = input("삭제할 이름: ")
            if name in address_book:
                del address_book[name]
                print(f"{name}님의 연락처가 삭제되었습니다.")
            else:
                print("등록되지 않은 이름입니다.")

        elif user == 3:
            name = input("검색할 이름: ")
            if name in address_book:
                print(f"{name}님의 전화번호: {address_book[name]}")
            else:
                print("등록되지 않은 이름입니다.")

        elif user == 4:
            if address_book:
                for name in sorted(address_book):
                    print(name, "의 전화번호:", address_book[name])
            else:
                print("등록된 연락처가 없습니다.")

        elif user == 5:
            print("종료합니다.")
            break

        else:
            print("1부터 5까지의 메뉴를 선택해주세요.")


def get_contact():
    name = input("이름: ")
    number = input("전화번호: ")
    print(f"이름: {name}")
    print(f"전화번호: {number}")
    return name, number


def display_menu():
    print("1. 연락처 추가")
    print("2. 연락처 삭제")
    print("3. 연락처 검색")
    print("4. 연락처 출력")
    print("5. 종료")

    try:
        return int(input("메뉴 항목을 선택하시오: "))
    except ValueError:
        return 0

main()
```
