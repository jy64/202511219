# 키를 이용한 정렬 예제
- 나이순으로 정렬하여 보여주는 프로그램

```python
class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __repr__(self):
        return "<이름: %s, 나이: %s>" % (self.name, self.age)

def kayAge(person):
    return person.age

people = [Person("홍길동", 20), Person("김철수", 35), Person("최자영", 38)]
print(sorted(people, key = kayAge))
```

***람다함수는 이름이 없는 함수를 만드는 함수,, def를 작성하지 않고 쓸 수 있다.(lambda 키워드 사용) 딱 한 번 사용되는 함수를 만들때 사용한다.***

# 람다식으로 온도 변환하기
- map()함수로 화씨 -> 섭씨로 변환하는 명령문을 작성

```python
f_temp = [0, 10, 20, 30, 40, 50]
c_temp = map(lambda x: (5.0/9.0)*(x-32.0), f_temp)
print(list(c_temp))
```

# 람다식으로 데이터 처리하기

-주문번호|물품|수량|품목 당 가격
---|---|---|---
1|"재킷"|5|120000
2|"셔츠"|6|24000
3|"바지"|3|50000
4|"코트"|6|300000

2개의 튜플이 저장된 리스트를 반환하는 프로그램 작성
각 튜플은 주문번호 및 총주문가격(수량*품목당가격)으로 구성
람다식 map()함수 사용

```python
orders = [["1","재킷",5,120000], ["2","셔츠",6,24000], ["3","바지",3,50000], ["4","코트",6,300000]]

result = list(map(lambda x: (x[0], x[2] * x[3]), orders))
print(result)
```

# 피보나치 이터레이터
- 앞의 두 수의 합이 바로 뒤의 수가 되는 수열을 의미
- __iter__ 와 __next__ 를 정의하여 클래스를 이터레이터 클래스로 작성

```python
class FibIterator:
    def __init__(self, a=1, b=0, maxValue=50):
        self.a = a
        self.b = b
        self.maxValue = maxValue

    def __iter__(self):
        return self

    def __next__(self):
        n = self.a + self.b
        if n > self.maxValue:
            raise StopIteration()
        self.a = self.b 
        self.b = n
        return n

for i in FibIterator():
    print(i, end=" ")
```
