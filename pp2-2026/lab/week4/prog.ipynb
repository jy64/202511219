```python
class Cat():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return self.name + "," + str(self.age)

    def setName(self, name):
        self.name = name

    def getName(self):
        return self.name

missy = Cat('Missy', 3)
lucky = Cat('Lucky', 5)

print(missy)
print(lucky)
```

```python
class Rocket():
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return "로켓의 위치: (" + str(self.x) + ", " + str(self.y) + ")"

    def moveUp(self):
        self.y = self.y + 1

myRocket = Rocket()
print("로켓의 높이:", myRocket.y)
myRocket.moveUp()
print("로켓의 높이:", myRocket.y)
```
