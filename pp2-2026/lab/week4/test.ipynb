```python
class MyClass:
    def __init__(self):
        self.radius = 10

    def set_radius(self, radius):
        self.radius = radius
```

```python
class MyClass:
    def __init__(self, r):
        self.__radius = r

    def get_radius(self):
        return self.__radius

a = MyClass(10)
print(a.get_radius())
```

```python
class Counter:
    count = 0
    def __init__(self):
        self.count = 1

c1 = Counter()
c2 = Counter()

print(c1.count)
```

```python
class Animal:
    def speak(self):
        return "Animal sound"

class Dog(Animal):
    def speak(self):
        return "bark!"

class Cat(Animal):
    pass

dog = Dog()
cat = Cat()

print(dog.speak() + "," + cat.speak())
```
