# 예제 : 사진 복사하기

```python
infile = open(r"C:\Users\박지연\Desktop\하치와레.jfif", "rb")
outfile = open("고양이.jfif", "wb")

while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()
print("하치와레.jfif를 " +"고양이.jfif"+"로 복사하였습니다. ")
```

# 작업폴더 : python 파일 실행하는 위치

```python
with open("file.txt", "r", encoding="utf-8") as f:
   # for line in f:
   #     print(line.strip())     
    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()
        
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
    
    f.seek(0,0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    
```
